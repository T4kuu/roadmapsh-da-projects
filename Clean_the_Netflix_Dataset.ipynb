Заменим NaN на строку "Unknown" т.к. это категории и не получится использовать среднии или медианные значения во избежание ошибочных данных

```python
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
df["date_added"] = df["date_added"].fillna("Unknown")
df["rating"] = df["rating"].fillna("Unknown")
df["duration"] = df["duration"].fillna("Unknown")
```

Разделим длительность на минуты для типа Movie и сезоны для типа TV Show создав 2 колонки "duration_value" и "duration_unit"

```python
df["duration_value"] = df["duration"].str.extract(r"(\d+)").astype("Int64")
df["duration_unit"] = df["duration"].str.extract(r"(min|Season|Seasons)")
print(df.head())
```

Преобразование колонки "release_year" в datetime не имеет смысла т.к. будут ошибочные данные для месяца и дня, оставим его в int32

```python
df["date_added"] = pd.to_datetime(df["date_added"], format="%B %d, %Y", errors="coerce")
df["release_year"] = pd.to_datetime(df["release_year"], format="%Y", errors="coerce")
df["release_year"] = df["release_year"].dt.year
print(df["date_added"].head())
print(df["release_year"].head())
```
