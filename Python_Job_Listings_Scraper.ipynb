```python
from bs4 import BeautifulSoup
import requests
import csv
```

```python
url = "https://realpython.github.io/fake-jobs/"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

job_title = soup.find_all("h2")
company_name = soup.find_all("h3")
location = soup.find_all("p", class_="location")
job_detail = soup.find_all("a", class_="card-footer-item", string="Apply")
```

```python
fields = ['Job Title', 'Company name', 'Location', 'Job Detail']

data = []

for i in range(100):
  data.append({
      'Job Title': job_title[i].text,
      'Company name': company_name[i].text,
      'Location': location[i].get_text(strip=True),
      'Job Detail': job_detail[i].get('href')
  })

filename = 'joblist.csv'

with open(filename, 'w') as csvfile:
  writer = csv.DictWriter(csvfile, fieldnames=fields)
  writer.writeheader()
  writer.writerows(data)
```

```python
with open('joblist.csv', mode='r') as file:
    csv_reader = csv.DictReader(file)
    data_read = []
    for row in csv_reader:
        data_read.append(row)

for d in data_read:
    print(d)
```
